#1.Fizz buzz

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

#2.Electricity bill calculator

```python
units = float(input('Enter the units of electricity consumed: '))

total_bill = 0

if units <= 100:
  total_bill = units * 5
elif units <= 200:
  total_bill = (100 * 5) + ((units - 100) * 7)
else:
  total_bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print(f'Your total electricity bill is: ₹{total_bill:.2f}')
```

#3.Display a pattern for a simple progress indicator.

```python
for i in range(1,6):
    for j in range(1,i+1):
        print("*",end="")
    print()
```

#4.palindrome

```python
t = input("Enter a word or phrase: ")
t = t.lower()
if t == t[::-1]:
 print("It is a palindrome.")
else:
 print("It is not a palindrome.")
```
